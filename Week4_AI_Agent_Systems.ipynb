```python

from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = Path("ai_ml_notes.txt").read_text(encoding="utf-8")

document = Document(
    page_content=text,
    metadata={"source": "ai_ml_notes.txt"}
)

splitter = RecursiveCharacterTextSplitter(
    chunk_size=250,
    chunk_overlap=40
)

chunks = splitter.split_documents([document])

print("Total chunks:", len(chunks))

for i, chunk in enumerate(chunks, start=1):
    print(f"\n--- Chunk {i} ---")
    print(chunk.page_content)
```

```python

from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# Convert text chunks into numerical embeddings
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Create FAISS vector database
vector_db = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)

print("FAISS vector database created successfully!")
print("Total documents stored:", vector_db.index.ntotal)
```

```python

question = "What is Deep Learning?"

results = vector_db.similarity_search(question, k=2)

print("Question:", question)
print("\nRetrieved information:")

for result in results:
    print("-", result.page_content)
```

```python

def rag_answer(question, k=2):
    # Step 1: Retrieve relevant chunks
    retrieved_docs = vector_db.similarity_search(
        question, k=k
    )

    # Step 2: Prepare context
    context = "\n\n".join(
        doc.page_content for doc in retrieved_docs
    )

    # Step 3: Generate a grounded answer
    # This baseline uses retrieved text, not an LLM.
    answer = (
        f"Question: {question}\n\n"
        f"Answer based on the retrieved documents:\n{context}"
    )

    return answer


question = "What is Deep Learning?"
print(rag_answer(question))
```

```python

from typing import TypedDict
from langgraph.graph import StateGraph, START, END

# Define workflow state
class RAGState(TypedDict):
    question: str
    context: str
    answer: str

# Node 1: Retrieve documents
def retrieve_node(state: RAGState):
    docs = vector_db.similarity_search(state["question"], k=2)

    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    return {"context": context}

# Node 2: Generate a baseline answer
def answer_node(state: RAGState):
    context = state["context"]

    answer = (
        "Based on the retrieved documents:\n" + context
        if context.strip()
        else "No relevant information was found in the documents."
    )

    return {"answer": answer}

# Build the graph
graph_builder = StateGraph(RAGState)

graph_builder.add_node("retrieve", retrieve_node)
graph_builder.add_node("generate_answer", answer_node)

graph_builder.add_edge(START, "retrieve")
graph_builder.add_edge("retrieve", "generate_answer")
graph_builder.add_edge("generate_answer", END)

rag_graph = graph_builder.compile()

print("LangGraph workflow created successfully!")
```

```python

result = rag_graph.invoke({
    "question": "What is Deep Learning?",
    "context": "",
    "answer": ""
})

print("Question:", result["question"])
print("\nAnswer:", result["answer"])
```

```python

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("Model loaded successfully on:", device)
```

```python

def generate_answer(prompt, max_new_tokens=100):
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens
        )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

print(generate_answer(
    "Explain RAG in two simple sentences."
))
```

```python
evaluation_results = []

for question in evaluation_questions:
    result = rag_graph.invoke({
        "question": question,
        "context": "",
        "answer": ""
    })

    answer = result["answer"]

    evaluation_results.append({
        "Question": question,
        "Answer": answer,
        "Answer Generated": bool(answer.strip())
    })

successful = sum(
    item["Answer Generated"] for item in evaluation_results
)

print("RAG EVALUATION REPORT")
print("=" * 40)
print("Total Questions:", len(evaluation_results))
print("Answers Generated:", successful)
print("Empty Answers:", len(evaluation_results) - successful)
print("Answer Generation Rate:", round(
    successful / len(evaluation_results) * 100, 2
), "%")
```
